```python
from Bio import SeqIO

def load_fasta(file_path):
    """Load sequences from a FASTA file."""
    sequences = []
    for record in SeqIO.parse(file_path, "fasta"):
        sequences.append(str(record.seq))  # Extract DNA sequence from each record
    return sequences

# Example usage:
file_path = '/Users/charandeepreddytanamala/Downloads/HW4.fas'  # Replace with the actual path to your file
sequences = load_fasta(file_path)

# Print the first few sequences to check
print("First few DNA sequences from the FASTA file:")
for seq in sequences[:5]:  # Show the first 5 sequences
    print(seq)
```

```python
import numpy as np

# Function to convert a DNA sequence into a feature vector (numerical)
def sequence_to_feature_vector(sequence):
    """Convert DNA sequence to a numerical feature vector."""
    mapping = {'A': 0, 'C': 1, 'G': 2, 'T': 3}
    return [mapping[base] for base in sequence]

# Step 2: Convert the sequences into numerical feature vectors
def convert_sequences_to_vectors(sequences):
    return np.array([sequence_to_feature_vector(seq) for seq in sequences])

# Convert sequences to numerical feature vectors
feature_vectors = convert_sequences_to_vectors(sequences)

# Print the first few feature vectors to check
print("First few feature vectors:")
for vec in feature_vectors[:5]:  # Show the first 5 feature vectors
    print(vec)
```

```python
from sklearn.cluster import KMeans

# Step 3: Perform K-means clustering
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(feature_vectors)

# Print the first few labels to check the clustering result
print("Cluster labels for the first few sequences:")
for label in labels[:5]:  # Show the first 5 cluster labels
    print(label)
```

```python
from sklearn.decomposition import PCA

# Step 4: Reduce dimensions to 2D using PCA for visualization
pca = PCA(n_components=2)
reduced_features = pca.fit_transform(feature_vectors)

# Print the first few reduced features to check
print("First few reduced features (PCA results):")
for vec in reduced_features[:5]:  # Show the first 5 reduced vectors
    print(vec)
```

```python
import matplotlib.pyplot as plt

# Step 5: Plot the clusters
plt.figure(figsize=(8, 6))
plt.scatter(reduced_features[:, 0], reduced_features[:, 1], c=labels, cmap='viridis', s=50)
plt.title("DNA Sequence Clustering (K-means)", fontsize=16)
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.colorbar(label='Cluster')
plt.show()
```

```python
import matplotlib.pyplot as plt

def plot_clusters(reduced_features, labels):
    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(reduced_features[:, 0], reduced_features[:, 1],
                          c=labels, cmap='viridis', s=50, edgecolor='k')
    plt.title("DNA Sequence Clustering (K-means)", fontsize=16)
    plt.xlabel("PCA Component 1")
    plt.ylabel("PCA Component 2")
    legend_labels = ['Cluster 0', 'Cluster 1', 'Cluster 2']
    handles = [plt.Line2D([], [], marker='o', linestyle='', color=scatter.cmap(scatter.norm(i)))
               for i in range(3)]
    plt.legend(handles, legend_labels)
    plt.grid(True)
    plt.show()
```

```python
plot_clusters(reduced_features, labels)
```
